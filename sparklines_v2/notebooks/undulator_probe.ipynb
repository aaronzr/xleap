# Undulator PV Archiver Probe

Run the recursive-bisect probe over every PV in the `undulator` group. For each PV, `bisect_until_success` halves the time window on every failed fetch, converging on the smallest window the archiver will answer. All helpers live in `sparklines_v2.archive.probe`.

```python
import datetime as dt
from pprint import pprint

from sparklines_v2.archive.probe import (
    DEFAULT_PROBE_MAX_DEPTH,
    DEFAULT_PROBE_MIN_WINDOW,
    DEFAULT_PROBE_TIMEOUT_SECONDS,
    find_pv_result,
    load_group_specs,
    run_undulator_probe,
    summarize_probe,
)

UNDULATOR_SPECS = load_group_specs("undulator")
UNDULATOR_SPECS
```

```python
START = dt.datetime(2025, 6, 18, 6, 0, 0)
END = dt.datetime(2025, 6, 18, 12, 0, 0)

probe_results = run_undulator_probe(
    START,
    END,
    timeout=DEFAULT_PROBE_TIMEOUT_SECONDS,
    min_window=DEFAULT_PROBE_MIN_WINDOW,
    max_depth=DEFAULT_PROBE_MAX_DEPTH,
)
pprint(summarize_probe(probe_results), sort_dicts=False)
```

```python
FOCUS_PV = "SIOC:SYS0:ML04:AO310"

focus_result = find_pv_result(probe_results, FOCUS_PV)
focus_result["attempts"]
```
